```python
from pathlib import Path

import pandas as pd

DATA = Path("..") / ".." / "data"
apps = pd.read_csv(DATA / "applications.csv", dtype=str)
products = pd.read_csv(DATA / "products.csv", dtype=str)
subs = pd.read_csv(DATA / "submissions.csv", dtype=str, parse_dates=["status_date"])
subs["submission_no"] = subs["submission_no"].astype(int)
```

## Question 9
TRAP QUESTION. How many submissions do NOT have review_priority 'STANDARD'? Write the obvious query first. Then ask: are the blank (NULL) priorities counted? Fix your query so you know exactly what you are counting.

```python
# the obvious version:
(subs["review_priority"] != "STANDARD").sum()
# the version that matches SQL's <> (ignores missing values):
(subs["review_priority"].notna() & (subs["review_priority"] != "STANDARD")).sum()
```
